```python
import pandas as pd
```

## create a dataframe views with two columns: datetime and user by reading feed- views.log 

```python
df = pd.read_csv('../../data/feed-views.log', sep='\t', names=['datetime', 'user'])
```

```python
df['datetime'] = pd.to_datetime(df['datetime'])
```

```python
df['datetime'].dtype
```

```python
df['year'] = df['datetime'].dt.year
```

```python
df['month'] = df['datetime'].dt.month
```

```python
df['day'] = df['datetime'].dt.day
```

```python
df['hour'] = df['datetime'].dt.hour
```

```python
df['minute'] = df['datetime'].dt.minute
```

```python
df['second'] = df['datetime'].dt.second
```

```python
df
```

## create the new column daytime

```python
# 0 – 3.59 night, 4 – 6.59 early morning, 7 – 10.59 morning, 11 – 16.59 afternoon, 17 – 19.59 early evening, 20 – 23.59 evening
```

```python
bins = [-0.1, 3.59, 6.59, 10.59, 16.59, 19.59, 23.59]
labels = ['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening']
```

```python
df['time_of_day'] = df['hour'] + df['minute'] / 100
```

```python
df
```

```python
df['daytime'] = pd.cut(df['time_of_day'], bins=bins, labels=labels)
```

```python
df = df.set_index('user')
```

```python
df
```
