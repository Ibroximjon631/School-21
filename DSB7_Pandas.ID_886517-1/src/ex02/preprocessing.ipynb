```python
import pandas as pd
```

## download and read the CSV file and make ID the index column

```python
df = pd.read_csv('auto.csv', index_col='ID')
```

```python
df
```

## count the number of observations using the method count()

```python
df.count()
```

## drop the duplicates, taking into account only the following columns: CarNumber, Make_n_model, Fines

```python
df = df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
```

```python
df.count()
```

## work with missing values

```python
df.isna().sum()
```

```python
df.shape
```

```python
df = df.dropna(thresh=224, axis=1)
```

```python
df.isna().sum()
```

```python
# df['Refund'] = df['Refund'].fillna(method='ffill')
df['Refund'] = df['Refund'].ffill()
```

```python
df.isna().sum()
```

```python
fines_mean = df['Fines'][df['Fines'] != 0].mean()
```

```python
df['Fines'] = df['Fines'].fillna(fines_mean)
```

```python
df.isna().sum()
```

## split and parse the make and model

```python
df['Make'] = df['Make_n_model'].apply(lambda x: x.split()[0])
```

```python
df['Model'] = df['Make_n_model'].apply(lambda x: x.split()[1] if len(x.split()) > 1 else None)
```

```python
df = df.drop('Make_n_model', axis=1)
```

```python
df.isna().sum()
```
