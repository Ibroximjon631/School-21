## iterations: in all the following subtasks, you need to calculate fines/refund*year for each row and create a new column with the calculated data and measure the time using the magic command %%timeit in the cell

```python
%%timeit -n1 -r1
## loop
list_with_values = list()
for i in range(len(df)):
    result = df.loc[i, 'Fines'] / df.loc[i, 'Refund'] * df.loc[i, 'Year']
    list_with_values.append(result)
df['New Column'] = list_with_values
print(df)
```

```python
%%timeit -n1 -r1
## iterrows()
list_with_values = list()
for index, row in df.iterrows():
    result = row['Fines'] / row['Refund'] * row['Year']
    list_with_values.append(result)
df['New Column'] = list_with_values
print(df)
```

```python
%%timeit -n1 -r1
## apply() and lambda function
df['New Column'] = df.apply(lambda x: x['Fines'] / x['Refund'] * x['Year'], axis=1)
print(df)
```

```python
%%timeit -n1 -r1
## using Series objects from the dataframe
df['New Column'] = df['Fines'] / df['Refund'] * df['Year']
print(df)
```

```python
%%timeit -n1 -r1
## method .values()
df['New Column'] = df['Fines'].values / df['Refund'].values * df['Year'].values
print(df)
```

## indexing: measure the time using the magic command %%timeit in the cell

```python
%%timeit -n1 -r1 
car_number = 'O136HO197RUS'
print(df[df['CarNumber'] == car_number])
```

```python
df = df.set_index('CarNumber')
```

```python
%%timeit -n1 -r1
car_number = 'O136HO197RUS'
print(df.loc[car_number])
```
