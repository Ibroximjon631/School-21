```python
import pandas as pd
import sqlite3
```

## make a connection to the database (it is the same as the previous day)


```python
connection = sqlite3.connect('../data/checking-logs.sqlite')
```

## run a query that gets the datetime from the pageviews table, selecting only the users and not the admins


```python
pd.read_sql('SELECT * FROM pageviews LIMIT 2', connection)
```

```python
df = pd.read_sql('SELECT datetime FROM pageviews WHERE uid LIKE "user%"', connection)
```

```python
df
```

## using Pandas, create a new dataframe where the visits are counted and grouped by date

```python
df['datetime'] = pd.to_datetime(df['datetime'])
```

```python
df['date'] = df['datetime'].dt.date
```

```python
df
```

```python
visits = df.groupby('date').size().reset_index(name='visits')
```

```python
visits
```

## using Pandas method .plot(), create a graph

```python
graph = visits.plot(kind='line', figsize=(15, 8), x='date', y='visits', title='Views per day', fontsize=8)
graph.tick_params(axis='x', rotation=90)
```

## close the connection to the database


```python
connection.close()
```
